```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid",palette="deep")
pd.set_option('display.max_columns',None)
print("All libraries imported successfully!")
```

```python
# Load the Tips Dataset

tips = sns.load_dataset('tips')

# Display first 5 rows
print("TIPS DATASET")
display(tips.head())
```

```python
# STEP 4: DATASET INFORMATION AND EXPLORATION

# Display Dataset Shape
print("Dataset Shape:")
print(tips.shape)

# Display Dataset Information
print("\nDataset Information:")
tips.info()

# Display Statistical Summary
print("\nStatistical Summary:")
display(tips.describe())

# Check Missing Values
print("\nMissing Values:")
print(tips.isnull().sum())

# Display Column Names
print("\nColumn Names:")
print(tips.columns.tolist())

print("\nDataset Exploration Completed Successfully!")
```

```python
# STEP 5: HISTOGRAM VISUALIZATION

# Create Histogram for Total Bill

plt.figure(figsize=(10, 6))

sns.histplot(data=tips, x='total_bill', bins=20, color='skyblue', edgecolor='black')

# Add Title and Labels
plt.title('Distribution of Total Restaurant Bills', fontsize=16)
plt.xlabel('Total Bill Amount (USD)', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)

# Display the Chart
plt.tight_layout()
plt.show()
```

```python
# STEP 6: BOX PLOT VISUALIZATION

# Create Box Plot for Tips by Day

plt.figure(figsize=(10, 6))

sns.boxplot(data=tips, x='day', y='tip', hue='day', palette='Set2', legend=False)

# Add Title and Labels
plt.title('Distribution of Tips Across Different Days', fontsize=16)
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Tip Amount (USD)', fontsize=12)

# Display the Chart
plt.tight_layout()
plt.show()
```

```python
# STEP 7: SCATTER PLOT VISUALIZATION

# Create Scatter Plot for Total Bill and Tip

plt.figure(figsize=(10, 6))

sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', palette='Set1', s=80)

# Add Title and Labels
plt.title('Relationship Between Total Bill and Tip Amount', fontsize=16)
plt.xlabel('Total Bill Amount (USD)', fontsize=12)
plt.ylabel('Tip Amount (USD)', fontsize=12)

# Display Legend
plt.legend(title='Time of Day')

# Display the Chart
plt.tight_layout()
plt.show()
```

```python
# STEP 8: CORRELATION HEATMAP VISUALIZATION

# Calculate Correlation Between Numerical Columns
correlation = tips.select_dtypes(include='number').corr()

# Create Heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(correlation, annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)

# Add Title
plt.title('Correlation Heatmap of Tips Dataset', fontsize=16)

# Display the Chart
plt.tight_layout()
plt.show()
```

```python
# STEP 9: BAR CHART VISUALIZATION

# Calculate Average Tips for Each Day
average_tips = tips.groupby('day', observed=True)['tip'].mean().reset_index()

# Create Bar Chart
plt.figure(figsize=(10, 6))

sns.barplot(data=average_tips, x='day', y='tip', hue='day', palette='Set2', legend=False)

# Add Title and Labels
plt.title('Average Tip Amount Across Different Days', fontsize=16)
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Average Tip Amount (USD)', fontsize=12)

# Display the Chart
plt.tight_layout()
plt.show()
```

```python
# STEP 10: PROFESSIONAL DATA VISUALIZATION DASHBOARD

# Create a 2x2 Dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Histogram
sns.histplot(data=tips, x='total_bill', bins=20, color='skyblue',
             edgecolor='black', ax=axes[0, 0])

axes[0, 0].set_title('Distribution of Total Bills', fontsize=14)
axes[0, 0].set_xlabel('Total Bill Amount (USD)')
axes[0, 0].set_ylabel('Number of Customers')

# 2. Box Plot
sns.boxplot(data=tips, x='day', y='tip', hue='day',
            palette='Set2', legend=False, ax=axes[0, 1])

axes[0, 1].set_title('Tips Distribution Across Days', fontsize=14)
axes[0, 1].set_xlabel('Day')
axes[0, 1].set_ylabel('Tip Amount (USD)')

# 3. Scatter Plot
sns.scatterplot(data=tips, x='total_bill', y='tip',
                hue='time', palette='Set1', s=60, ax=axes[1, 0])

axes[1, 0].set_title('Relationship Between Bill and Tip', fontsize=14)
axes[1, 0].set_xlabel('Total Bill Amount (USD)')
axes[1, 0].set_ylabel('Tip Amount (USD)')

# 4. Correlation Heatmap
correlation = tips.select_dtypes(include='number').corr()

sns.heatmap(correlation, annot=True, cmap='Blues',
            fmt='.2f', linewidths=0.5, ax=axes[1, 1])

axes[1, 1].set_title('Correlation Heatmap', fontsize=14)

# Add Main Dashboard Title
fig.suptitle('TIPS DATASET - DATA VISUALIZATION DASHBOARD',
             fontsize=20, fontweight='bold')

# Adjust Layout
plt.tight_layout(rect=[0, 0, 1, 0.96])

# Save Dashboard Image
plt.savefig('professional_dashboard.png', dpi=150, bbox_inches='tight')

# Display Dashboard
plt.show()

print("Professional Dashboard Created Successfully!")
print("Dashboard saved as professional_dashboard.png")
```
